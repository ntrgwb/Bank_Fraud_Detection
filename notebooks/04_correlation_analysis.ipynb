```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/processed/base_cleaned.csv")

df.shape
```

```python
numeric_df = df.select_dtypes(include="number")

numeric_df.shape
```

```python
fraud_corr = (
    numeric_df.corr()["fraud_bool"]
    .drop("fraud_bool")
    .sort_values()
)

fraud_corr
```

```python
fraud_corr_abs = fraud_corr.abs().sort_values(ascending=False)

fraud_corr_abs
```

```python
top15 = fraud_corr_abs.head(15)

plt.figure(figsize=(10, 6))

plt.barh(
    top15.index[::-1],
    top15.values[::-1]
)

plt.xlabel("Absolute Correlation with Fraud")
plt.ylabel("Feature")
plt.title("Top 15 Numerical Features Correlated with Fraud")
plt.grid(axis="x", alpha=0.3)

plt.show()
```
